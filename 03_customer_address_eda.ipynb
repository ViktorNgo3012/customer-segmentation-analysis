## State

This section examines the geographical distribution of customers across Australian states.

Customer locations are analysed to understand the regional composition of the existing customer base.

```python
# Display state frequency

address["state"].value_counts()
```

```python
# Display state percentage

address["state"].value_counts(normalize=True).mul(100).round(2)
```

```python
# Plot state distribution

state_counts = address["state"].value_counts()

state_counts.plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("Distribution of Customer State")
plt.xlabel("State")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()
```

### Observation

- Customers are distributed across **three Australian states**, with **New South Wales (NSW)** representing the largest proportion of the existing customer base.
- **NSW** accounts for **53.51% (2,140 customers)**, followed by **Victoria (VIC)** with **25.53% (1,021 customers)** and **Queensland (QLD)** with **20.96% (838 customers)**.
- The bar chart shows that more than half of all customers are located in **NSW**, while **VIC** and **QLD** contribute approximately one-quarter and one-fifth of the customer base, respectively.
- Overall, the cleaned customer address dataset indicates that the existing customer base is geographically concentrated in **New South Wales**, with additional customer representation across **Victoria** and **Queensland**, providing regional diversity for future customer segmentation and location-based marketing strategies.

## Property Valuation

This section examines the distribution of customer property valuation scores.

Descriptive statistics and visualisations are used to understand property valuation patterns, variability, and potential outliers.

```python
# Display summary statistics for property valuation

address["property_valuation"].describe()
```

```python
# Plot distribution of property valuation

plt.figure(figsize=(8,4))

plt.hist(
    address["property_valuation"],
    bins=12
)

plt.title("Distribution of Property Valuation")
plt.xlabel("Property Valuation")
plt.ylabel("Frequency")

plt.grid(True)

plt.show()
```

```python
# Plot boxplot of property valuation

plt.figure(figsize=(8,2.5))

plt.boxplot(
    address["property_valuation"],
    vert=False
)

plt.title("Boxplot of Property Valuation")
plt.xlabel("Property Valuation")

plt.grid(True)

plt.show()
```

### Observation

- Property valuation scores range from **1 to 12**, indicating variation in customer property values across the existing customer base.
- The **average property valuation score is 7.51**, while the **median score is 8**, suggesting that most customers have moderate to relatively high property valuation scores.
- The middle **50%** of customers have property valuation scores between **6 and 10**, indicating moderate variability in property values.
- The histogram shows that customer property valuation scores are concentrated around the middle to higher valuation levels, with fewer customers at the lowest valuation scores.
- The boxplot indicates **no apparent extreme outliers**, suggesting that property valuation scores fall within the expected range of the scoring system.
- Overall, the cleaned customer address dataset indicates that most customers are associated with moderate to high property valuation scores, providing a useful attribute for future customer segmentation and profiling.

## Postcode

This section examines the most common customer postcodes.

Because the dataset contains a large number of unique postcodes, only the top 10 most frequent postcodes are presented to improve readability and highlight the areas with the highest customer concentrations.

```python
# Display top 10 postcodes

address["postcode"].value_counts().head(10)
```

```python
# Display percentage of top 10 postcodes

address["postcode"].value_counts(normalize=True).mul(100).head(10).round(2)
```

```python
# Plot top 10 postcodes

postcode_counts = address["postcode"].value_counts().head(10)

postcode_counts.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top 10 Customer Postcodes")
plt.xlabel("Postcode")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()
```

### Observation

- The customer address dataset contains a large number of unique postcodes, indicating that customers are geographically distributed across many locations.
- The most common postcode is **2170**, representing **0.78% (31 customers)** of the customer base.
- Other frequently occurring postcodes include **2145 (0.75%, 30 customers)**, **2155 (0.75%, 30 customers)**, **2153 (0.73%, 29 customers)**, **3977 (0.65%, 26 customers)**, **2560 (0.65%, 26 customers)**, **2770 (0.65%, 26 customers)**, **2065 (0.63%, 25 customers)**, **2250 (0.55%, 22 customers)**, and **2148 (0.55%, 22 customers)**.
- The bar chart shows that no individual postcode contains a substantial proportion of customers, with even the most common postcode accounting for less than **1%** of the customer base.
- Overall, the cleaned customer address dataset demonstrates a geographically dispersed customer base, with customers spread across numerous postcodes rather than concentrated within a small number of locations.
