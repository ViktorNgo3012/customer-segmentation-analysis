The objective is to understand the customer base and identify demographic patterns that may support future customer segmentation and marketing strategies.

## Observation

- The cleaned customer demographic dataset contains **4,000 customer records** and **9 variables**, covering customer demographics, purchasing history, employment information, vehicle ownership, wealth segment, customer tenure, and age.
- As the dataset has already undergone the data cleaning process, it contains **no missing values** and **no duplicate records**, making it suitable for exploratory data analysis.
- The dataset contains **four numerical variables** (`customer_id`, `past_3_years_bike_related_purchases`, `tenure`, and `age`) and **five categorical variables** (`gender`, `job_title`, `job_industry_category`, `wealth_segment`, and `owns_car`).
- Customers are represented across **196 unique job titles** and **10 industry categories**, providing rich demographic information for customer profiling and segmentation.
- Overall, the cleaned dataset is complete, consistent, and well-prepared for analysing customer characteristics and identifying demographic patterns that support customer segmentation.

## Gender

This section examines the distribution of customer gender.

Customer gender is analysed to understand the demographic composition of the existing customer base.

```python
# Display gender frequency

customers["gender"].value_counts()
```

```python
# Display gender percentage

customers["gender"].value_counts(normalize=True).mul(100).round(2)
```

```python
# Plot gender distribution

gender_counts = customers["gender"].value_counts()

gender_counts.plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("Distribution of Customer Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()
```

## Observation

- The customer base is almost evenly split between **Female** and **Male** customers, with females representing **50.98% (2,039 customers)** and males representing **46.82% (1,873 customers)** of the dataset.
- Only **2.20% (88 customers)** are recorded as **Unknown**, indicating that the vast majority of customer gender information is available.
- The bar chart shows a relatively balanced gender distribution, with only a small difference between the number of female and male customers.
- Overall, the customer demographic dataset exhibits a well-balanced gender composition, suggesting that both female and male customers are similarly represented within the existing customer base.

## Age

This section examines the distribution of customer age.

Descriptive statistics and visualisations are used to understand customer age patterns, variability, and potential outliers.

```python
# Display summary statistics for age

customers["age"].describe()
```

```python
# Plot age distribution

plt.figure(figsize=(8,4))

plt.hist(
    customers["age"],
    bins=20
)

plt.title("Distribution of Customer Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.grid(True)

plt.show()
```

```python
# Plot boxplot of age

plt.figure(figsize=(8,2.5))

plt.boxplot(
    customers["age"],
    vert=False
)

plt.title("Boxplot of Customer Age")
plt.xlabel("Age")

plt.grid(True)

plt.show()
```

### Observation

- Customer ages range from **24 to 182 years**, indicating substantial variation in the ages of customers within the existing customer base.
- The average customer age is **48.29 years**, while the median age is **48 years**, suggesting that the age distribution is approximately symmetric, although influenced by several high-age observations.
- The middle **50%** of customers are between **39 and 57 years** of age, demonstrating that most customers belong to the middle-aged demographic.
- The histogram indicates that customer ages are concentrated primarily between **35 and 65 years**, with relatively few customers outside this range.
- The boxplot shows **several high-age outliers**, including an extreme value of **182 years**, which is unlikely to represent a realistic customer age and may indicate a remaining data quality issue despite the data cleaning process completed earlier.
- Overall, the cleaned customer demographic dataset is dominated by middle-aged customers, with only a small number of unusually high age values that should be interpreted cautiously during further analysis.
