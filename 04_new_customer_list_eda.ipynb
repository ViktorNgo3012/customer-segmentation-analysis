## Past 3 Years Bike Related Purchases

The distribution of customers' bike-related purchases over the past three years was examined to understand previous purchasing behaviour among prospective customers.

```python
# Summary statistics

new_customers["past_3_years_bike_related_purchases"].describe()
```

```python
# Histogram

plt.figure(figsize=(8,5))

plt.hist(
    new_customers["past_3_years_bike_related_purchases"],
    bins=10
)

plt.title("Distribution of Past 3 Years Bike Related Purchases")
plt.xlabel("Number of Purchases")
plt.ylabel("Frequency")

plt.show()
```

```python
# Box plot

plt.figure(figsize=(8,2))

plt.boxplot(
    new_customers["past_3_years_bike_related_purchases"],
    vert=False
)

plt.title("Box Plot of Past 3 Years Bike Related Purchases")
plt.xlabel("Number of Purchases")

plt.show()
```

```python
# Check for outliers using IQR

Q1 = new_customers["past_3_years_bike_related_purchases"].quantile(0.25)
Q3 = new_customers["past_3_years_bike_related_purchases"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = new_customers[
    (new_customers["past_3_years_bike_related_purchases"] < lower_bound) |
    (new_customers["past_3_years_bike_related_purchases"] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

### Observation

The number of bike-related purchases made by prospective customers over the past three years ranged from **0 to 99**, with an average of **49.84 purchases** (SD = **27.80**). The median was **51 purchases**, indicating that purchasing behaviour was relatively balanced around the centre of the distribution.

The first quartile (Q1 = **26.75**) and third quartile (Q3 = **72.00**) indicate that the middle 50% of customers made between approximately **27 and 72 purchases** during the three-year period. Using the IQR method, no outliers were identified, as all observations fell within the calculated lower (-41.13) and upper (139.88) bounds.

Overall, the distribution suggests considerable variation in customers' historical purchasing behaviour, ranging from customers with no previous purchases to highly active purchasers. The absence of outliers indicates that all purchase frequencies fall within the expected range and can be retained for subsequent customer segmentation and purchasing behaviour analysis.

## Correlation Analysis

This section examines the linear relationships between the numerical variables in the prospective customer dataset.

A Pearson correlation matrix is used to assess the direction and strength of relationships among bike-related purchases, tenure, property valuation, and age. Postcode is excluded because it is a geographic identifier rather than a continuous numerical measure.

Correlation values range from **-1 to 1**:

- Values close to **1** indicate a strong positive linear relationship.
- Values close to **-1** indicate a strong negative linear relationship.
- Values close to **0** indicate little or no linear relationship.

Correlation identifies association rather than causation.

```python
# Select meaningful numerical variables

correlation_variables = [
    "past_3_years_bike_related_purchases",
    "tenure",
    "property_valuation",
    "age"
]

correlation_matrix = new_customers[correlation_variables].corr()

correlation_matrix.round(2)
```

```python
import seaborn as sns
```

```python
# Plot correlation matrix

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()

plt.show()
```

### Observation

The correlation analysis indicates that most numerical variables have little to no linear relationship with one another. The strongest correlation is observed between **tenure** and **age** (r = 0.34), suggesting a weak positive relationship where older prospective customers tend to have slightly longer customer tenure. However, the correlation is not strong enough to indicate a substantial linear association.

Past 3 years bike-related purchases exhibit negligible correlations with tenure (r = -0.03), property valuation (r = -0.01), and age (r = -0.02), indicating that previous purchasing behaviour is largely independent of these variables in the prospective customer dataset. Similarly, property valuation has almost no linear relationship with either tenure (r = -0.01) or age (r = -0.04).

Overall, the low correlation coefficients suggest limited multicollinearity among the numerical variables, meaning each variable is likely to contribute distinct information for subsequent customer segmentation and analysis.
